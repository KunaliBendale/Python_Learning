```python
try:
    num = int(input("Enter number: "))
    if num<=0:
        raise ValueError("invalid number")
    else:
        square_root=num**0.5
        print(square_root)
except Exception:
    print('Invalid Literal')
```

```python
# WAP to access element from list using an index entered by user , handle IndexError

list1=["kunali","shamal","rutuja","yash"]

try:
    index=int(input("Enter index to access element from list: "))
    if index>=len(list1):
        raise IndexError("index out of bound")
    else:
        print(list1[index])
except Exception as e:
    print(e)
```

```python
# WAP to access value from dictionary using key entered by user,handle KeyError

dict1={
    "name":'kunali',
    "age":22,
    "city":"pune" 
}

try:
    key=input("Enter key: ")
    if key not in dict1:
        raise KeyError("KeyError")
    else:
        print(dict1.get(key))
except Exception as e:
    print(e)
```

```python
# WAP that converts a string into integer. handle invalid conversion

try:
    str1=input("Enter string: ")
    conversion=int(str1)
    print(type(conversion))
except Exception:
    print("invalid conversion")
```

#User Defined Exception OR Custom EXCEPTION

```python

    
class NegativeAge(Exception):
    pass

age=int(input("ENter your age: "))
if age>=18:
    print("Eligible to vote")
elif age<0:
    raise NegativeAge("Invalid Age")
else:
    print("Not eligible to vote")
```

```python
class NegativeAge(Exception):
    pass

try:
    age=int(input("ENter your age: "))
    if age>=18:
        print("Eligible to vote")
    elif age<0:
        raise NegativeAge("Invalid Age")
    else:
        print("Not eligible to vote")
except NegativeAge:
    print("Invalid Age")
    
```

```python
class InsufficientBalanceError(Exception):
    pass

acc_balance=100000
withdrawal_amt=int(input("Enter withdrawal amount: "))

if withdrawal_amt>acc_balance:
    raise InsufficientBalanceError("Insufficient Balance..")
else: 
    print("Amount withdrawn successfully !!")
    print("Remaining balance is: ",acc_balance-withdrawal_amt)      
```

```python
class InvalidMarksError(Exception):
    pass

marks=int(input("Enter your marks: "))
if marks<0 or marks >100:
    raise InvalidMarksError("Invalid Marks")
else:
    print("valid marks")
```
