```python
Descriptors= :
    It is any object that implements at least on method - 
    1) __get__()  => called when retrive the value of attribute
    2) __set__()  => called when attribute managed by descriptor is assigned new value 
                        by allowing user to customize or enforce rules for assignment
    3) __delete__() => useful for managing resources, cleaning up when attributes is removed.
    
```

```python
class Student:
    def getAge(self):
        print("getter is called")
        return self._age
    def setAge(self,age):
        self._age=age
        print("setter is called..")
    def delAge(self):
        del self._age
        print("Deleter is called..")
    
    age=property(getAge,setAge,delAge)     # Descriptor using property method
    
s=Student()
s.age=25
print(s.age)
del s.age
```

```python
#Descriptor using annotations

class Person:
    
    @property
    def age(self):
        print("Getter is called..")
        return self._age
    
    @age.setter
    def age(self,age):
        self._age=age
        print("Setter is called")
    
    @age.deleter
    def age(self):
        del self._age
        print("deleter is called")
    
p=Person()
p.age=25
print(p.age)
del p.age
```

```python
# create a descriptor age to validate age

class Person:
    
    @property
    def age(self):
        return self._age
    
    @age.setter
    def age(self,age):
        if type(age)==int:
            if age >0 and age <100:
                self._age=age
            else:
                print("invalid age")
                self._age=None
    
    @age.deleter
    def age(self):
        del self._age
    
p=Person()
p.age=18
print(p.age)
del p.age
```

```python
class Person:
    
    @property
    def name(self):
        return self._name
    
    @name.setter
    def name(self,name):
        if type(name)==str:
            if len(name)<=8:
                self._name=name
            else:
                print("invalid name")
                self._name=None
    
    @name.deleter
    def name(self):
        del self._name
    
p=Person()
p.name="kunali"
print(p.name)
#del p.name
```
