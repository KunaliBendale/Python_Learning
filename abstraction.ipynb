# Abstraction

It means hiding the implementation from users and accessible only essential features

```python
#objects are only created of subclasses


from abc import ABC,abstractmethod

# ABC - Abstract Base Class --> used to make class Abstract class
# abstractmethod -  used to make abstractmethod

class Animal(ABC):              # abstract class --> contains abstract methods and attributes
    
    @abstractmethod           
    def Speak(self):             # Abstract method --> method without implementation
        pass
    
    @abstractmethod
    def Movement(self):
        pass

class Dog(Animal):              # Subclass of abstract class
    
    def Speak(self):            # Concrete method --> method with implementation
        return "Bark"
    def Movement(self):
        return "Running"
    
class Fish(Animal):
    def Speak(self):
        return "Blub"
    def Movement(self):
        return "Swimming"
    
    
dog1=Dog()
print(dog1.Speak())
print(dog1.Movement())

fish1=Fish()
print(fish1.Speak())
print(fish1.Movement())
```

```python
#3 

from abc import ABC,abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
        
    @abstractmethod
    def stop(self):
        pass
    
class Car(Vehicle):
    def start(self):
        return "Car is started"
    
    def stop(self):
        return "Car is stopped"

C1=Car()
print(C1.start())
print(C1.stop())
```

```python
#4
from abc import ABC ,abstractmethod

class Shape(ABC):
    
    @abstractmethod
    def Draw(self):
        pass

class Circle(Shape):
    def Draw(self,color,size):
        print("color of circle is :",color)
        print("Size of Circle (radius) is: ",size)

c1=Circle()
c1.Draw("blue",5)
```

```python
#5 
from abc import ABC , abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def area(self,radius):
        return 3.14*radius*radius
    
class Square(Shape):
    def area(self,side):
        return side*side
    
        
c1=Circle()
print("Area of circle is :",c1.area(5))

s1=Square()
print("Area of square is :",s1.area(5))
```

```python
#6

from abc import ABC,abstractmethod

class Employee(ABC):
    basicSalary=10000
    
    @abstractmethod
    def calculate_salary(self):
        pass
    
class FullTimeEmployee(Employee):
    def calculate_salary(self):
        return self.basicSalary

class PartTimeEmployee(Employee):
    def __init__(self,hrs,pay_per_hr):
        self.hrs=hrs
        self.pay_per_hr=pay_per_hr
        
    def calculate_salary(self): 
        return self.hrs * self.pay_per_hr
        
    

FE=FullTimeEmployee()
print("Your salary is: ",FE.calculate_salary())

PE=PartTimeEmployee(5,1000)
print("Your salary is: ",PE.calculate_salary())
```

```python
#8
from abc import ABC,abstractmethod

class Database(ABC):
    @abstractmethod
    def connect(self):
        pass
    @abstractmethod
    def disconnect(self):
        pass
    @abstractmethod
    def execute_query(self):
        pass
    
class MySQLDatabase(Database):
    
    def connect(self):
        return "Database Connected"
    
    def disconnect(self):
        return "Database Disconnected"
    
    def execute_query(self):
        return "Select * from DB_name"

class MongoDBDatabase(Database):
    def connect(self):
        return "use connect() method to sonnect with database"
    def disconnect(self):
        return "Databse disconnected"
    def execute_query(self):
        return "used find() method to fetch data from db"
    
MySQL = MySQLDatabase()
print(MySQL.connect())
print(MySQL.disconnect())
print(MySQL.execute_query())

mongo=MongoDBDatabase()
print(mongo.connect())
print(mongo.disconnect())
print(mongo.execute_query())
```
