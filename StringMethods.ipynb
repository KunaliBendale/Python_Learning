```python
text="String Methods are used to manipulate String !"

print(len(text))
print(text.upper())
print(text.lower())

print(text.capitalize())    # Converts only first letter of string in upper case
print(text.swapcase())
print(text.title())        #Converts each first letter of a word in upper case 
print(text.istitle())
print(text.isalpha())      # checks whether string contain alphabets, without space and digits
print(text.isalnum())      # checks whether string contain alphabets and digits without space.
print(text.startswith("S"))
print(text.endswith("!"))
print(text.count("e"))      # return no. of occurence of character
print(text.find("a"))
print(text.rfind("M"))      #rfind() checks charcter from right to left
print(text.replace("String","List"))
print(text.strip())
print(text.isspace())


print(sorted(text))
```

```python
# String Formatting

name="Kunali"
surname="Bendale"

str1=f"My name is {name} and surname is {surname}"     # f-string
print(str1)

str2= "My name is {} and surname is {}".format(name,surname)    # . format method
print(str2)
```
