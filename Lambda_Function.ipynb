```python
# Lambda Function - 
# It is anonymous function without having any name.

Questions :
1. Write a lambda function that adds 10 to a given number.
2. Convert a list of strings to uppercase.
3. Add corresponding elements of two lists.
4. Filter names that start with 'A'.
5. Filter out numbers greater than 10.
6. Write a lambda function that takes two parameters and returns their product.
Call the lambda function with the values 5 and 10.
7. Use a lambda function to square each number in the list [1, 2, 3, 4].
Print the results.
8. Use a lambda function to filter out all even numbers from the list [1, 2, 3, 4, 5, 6].
Print the results.
9. Remove Empty Strings
10. Sort Words by Length
11.Extract Last Digit of Each Number
12. Find Max String by Length
13. Use a lambda function as the key argument to sort a list of tuples by the second element.

Example input: [(1, 3), (2, 2), (3, 1)]
Output: [(3, 1), (2, 2), (1, 3)]
```

```python
# 1. Write a lambda function that adds 10 to a given number.

add_no=lambda x : x+10
add_no(5)
```

```python
# 2. Convert a list of strings to uppercase.
l1=["kunali","hello","bye"]
convert_to_upper=list(map(lambda item : item.upper(),l1))

print(convert_to_upper)
```

```python
#3. Add corresponding elements of two lists.

list1=[2,3,4,5,67]
list2=[9,7,6,8,4]
result=list(map(lambda x,y: x + y,list1,list2))
print(result)
```

```python
# 4. Filter names that start with 'A'.
list1=["APple","Amla","Banana","Avocado"]

result = list(filter(lambda x:x.startswith('A'),list1))
print(result)
```

```python
# 5. Filter out numbers greater than 10.
numbers=[3,6,9,45,23,7,43,90]

result=list(filter(lambda x: x>10,numbers))
print(result)
```

```python
# 6. Write a lambda function that takes two parameters and returns their product.

prod=lambda x,y : x*y
print(prod(5,10))
```

```python
# 7. Use a lambda function to square each number in the list [1, 2, 3, 4].
#Print the results.

list1= [1, 2, 3, 4]

result=list(map(lambda x:x*x,list1))
print(result)
```

```python
#8. Use a lambda function to filter out all even numbers from the list [1, 2, 3, 4, 5, 6].
#Print the results.

list1=[1, 2, 3, 4, 5, 6]

result=list(filter(lambda x: x%2==0,list1))
print(result)
```

```python
#9. Remove Empty Strings
list_of_str=["kunali","","Yash","",""]

result=list(filter(lambda x:x != "",list_of_str))
print(result)
```

```python
#10. Sort Words by Length

# sorted(iterable, key=function)

list_of_str=["kunali","Bhumika","Yash","om","suyash"]

result= sorted(list_of_str,key=lambda name : len(name))
print(result)
```

```python
# 11.Extract Last Digit of Each Number
numbers=[345,796,98,686,466]
result=list(map(lambda x: x%10,numbers))
print(result)
```

```python
# 12. Find Max String by Length

list_of_str=["kunali","Bhumika","Yash","om","suyash"]

max(list_of_str,key=lambda x:len(x))
```

```python
#13. Use a lambda function as the key argument to sort a list of tuples by the second element.

list1=[(1, 3), (2, 2), (3, 1)]
#Output: [(3, 1), (2, 2), (1, 3)]

result=sorted(list1,key=lambda item : item[1])
print(result)
```
