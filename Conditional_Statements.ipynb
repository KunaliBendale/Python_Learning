```python
# if-else statement

age=int(input("ENter your age: "))
if age>=18:
    print("You are eligible for voting..")
else:
    print("you are not eligible for voting..")
```

```python
# if-elif-else exmple 
num1=int(input("enter 1st number: "))
num2=int(input("Enter 2nd number: "))

if num1>num2 :
    print(f"{num1} is greater number")
elif num2>num1:
    print(f"{num2} is greater number")
else:
    print("Both numbers are same..")
```

```python
num1=int(input("Enter the number: "))
if num1>0:
    print("Number is positive")
elif num1<0:
    print("Number is negative")
else:
    print("number is 0") 
```

```python
# accept string from user and check whether the string is palindrome or not

str1=input("Enter string")

if str1== str1[::-1]:
    print("String is palindrome")
else:
    print("String is not palindrome")
```

```python
# Accept single character from user, and check whether it is vowelor consonent.

character=input("Enter single character: ")
if character in "AEIOUaeiou":
    print("Given character is vowel")
else:
    print("Given character is consonent")
```

```python
# accept no from user and check even or odd

num1=int(input("ENter number: "))
if num1 % 2==0:
    print("Number is even")
else:
    print("Number is odd")
```

```python
#accept 3 nos from user and print largest among them

num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))
num3 = int(input("Enter number 3: "))

if num1 > num2 and num1 > num3:
    print(f"{num1} is greater than {num2} and {num3}")
elif num2 > num1 and num2 > num3:
    print(f"{num2} is greater than {num1} and {num3}")
else:
    print(f"{num3} is greater than {num1} and {num2}")
```
