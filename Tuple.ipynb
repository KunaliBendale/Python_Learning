```python
# To modify tuple, we need to do type casting (tuple->list)
tup1=(23,45,67,78,45,89)
print("Tuple: ",tup1)
list1=list(tup1)
list1.append(90)
tup1=tuple(list1)
print("modified tuple: ",tup1)
```

```python
#4. find max and min value from tuple without using max and min

tup1=(12,34,1,45,2,66,8)
t2=sorted(tup1)
print(t2)
print("minimum element:",t2[0])
print('maximum element:',t2[-1])
```
