```python
# Calculate area of circle

PI=3.14
radius=int(input("Enter radius of circle :"))
area=PI * radius**2
print("Area of circle is :",area)
```

```python
# Calculate area of rectangle
length=int(input("Enter length of rectangle "))
breadth= int(input("Enter breadth of rectangle "))
print("Area of rectangle is :",length*breadth)
```

```python
#Calculate perimeter of rectangle
length=int(input("Enter lenght of rectangle to calculate perimeter "))
breadth=int(input("Enter breadth of rectangle to calculate perimeter "))
print("The perimeter of rectangle is :",(2*(length+breadth)))
```

```python
# Calculate area of triangle

base=int(input("Enter base of triangle "))
height=int(input("enter height of triangle "))

print("The area of triangle is ", (1/2)*base*height)
```

```python
# Convert tempreature from celsius to fahrenheit and vice versa

celsius=int(input("Enter temperature in celsius"))

Fahrenheit = (celsius*9/5) + 32 
print("Temperature in Fahrenheit : ", Fahrenheit)

FahrenheitT = int(input("Enter temperature in fahrenheit "))
CelsiusT = (FahrenheitT - 32) * 5/9
print("Temperature in Celsius : ", CelsiusT)
```

```python
str1=input("Enter the string : ")  

# 1. print first character of string
print("first character of string ",str1[0])

#2. print last character of string
print("last character of string ",str1[-1])

#3. print first 5 characters of string
print("first 5 characters ",str1[0:6])

#4. reverse string using slicing
print("reverse string using slicing ",str1[::-1])

#5. print every 2nd character from string
print("printed every secong charcater from string ",str1[::2])

#6. print first and last character of string using slicing
print("first and last character: ",str1[0],str1[-1])

#7. print everything except first and last character
print("except first and last charcater: ",str1[1:-1])

#8. print middle index character
length=len(str1)
print("middle index character: ",str1[length//2])

#9. divide string into two halves and print both

print("First Half: ",str1[0:len(str1)//2])
print("Secong Half: ",str1[len(str1)//2:])

#10.  replace first character with @
print("replacing first character with'@' :","@"+ str1[1:])

#11. print every 3rd character starting from 0
print("printing every 3rd character: ",str1[0::3])
```

```python
# check string is palindrome

str2=input("Enter the string ")
if str2 == str2[::-1]:
    print("string is palindrome")
else:
    print("string is not palindrome")

```

```python
# extract characters from index 2 to 8 with 2 step

text=input("enter string : ")
print("characters from index 2 to 8 with step 2: ",text[2:9:2])
```

```python
#14. swapping first and secong halves of string 
str12=input("enter string: ")
print("swapping first and secong halves of string: ",str12[len(str12)//2:] + str12[0:len(str12)//2] )

#15. Create a new string with characters only at odd indices.
print("New string with Characters only at odd indices : ", str12[0:len(str12):2])

#16. Extract last 4 characters of string without using len()
#str12[-1:-5]
print("Extracted last 4 characters: ",str12[-4:])

#17. reverse portion of string from index 2 to 7
newstr=str12[2:8]
print("After reversing portion of string from index 2 to 7: ",newstr[::-1])
```
