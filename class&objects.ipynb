```python
class Dog:
    species="caniene"
    
    def __init__(self,Name,Age):
        self.Name=Name
        self.Age=Age
    
    def info(self):
        return f"Name of dog is {self.Name} and Age is {self.Age}"

dog1=Dog("Buddy",2)
print(dog1.Name)
print(dog1.Age)
print(dog1.info())
print(dog1.species)
print("--------------------------------------")

dog2=Dog("Tommy",3)
print(dog2.Name)
print(dog2.Age)
print(dog2.info())
```

```python
class employee:
    company_Name="TCS"
    
    def __init__(self,Role,Salary):
        self.Role=Role
        self.Salary=Salary
    def details(self):
        print(f"The Employee belongs to {employee.company_Name} with role {self.Role} and salary {self.Salary} !!")
        
emp1=employee("Software Enginner",100000)
emp1.details()

emp2=employee("Software Tester",50000)
emp2.details()
```

```python
#7 create class Car with 2 constructors , one for setting make and model and another for make ,model and year

class Car:
        
    def __init__(self,make,model,year=None):
        self.make=make
        self.model=model
        self.year=year
    
    def info(self):
        if self.year:
            print(f"Car Make {self.make} model is {self.model} year is {self.year}")
        else:
            print(f"Car Make {self.make} model is {self.model}")
        
c1=Car("Tata","Nexon",2020)
c1.info()
c2=Car("Tata","Nexon")
c2.info()
    
```

```python
#8. Write a Python program that defines a Person class with a constructor that takes age as an argument. 
#If the age is less than 18, assign the person to a "minor" group; otherwise, assign them to an "adult" group.

class Person:
    def __init__(self,age):
        self.age=age
        
    def check(self):
        if self.age<18:
            print("group is minor")
        else:
            print("group is adult")
            
p1=Person(12)
p1.check()

p2=Person(23)
p2.check()
```
