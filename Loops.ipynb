```python
# separate odd and even nos, from list
numbers=[23,12,56,43,78,79,5,37,89,80,55]
even_nos=[]
odd_nos=[]

for num in numbers:
    if num % 2 == 0:
        even_nos.append(num)
    else:
        odd_nos.append(num)
print("Even numbers list: ",even_nos)
print("Odd numbers list: ",odd_nos)
```

```python
# separate vowels and consonents into two list

elements = ['a','b','c','e','f','T','h','n','i','o','p','Z']
vowels_list=[]
consonant_list=[]

for element in elements:
    if element in 'aeiouAEIOU':
        vowels_list.append(element)
    else:
        consonant_list.append(element)
        
print("Vowels list: ",vowels_list)
print("COnsonant list: ",consonant_list)
```
