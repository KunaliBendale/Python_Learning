```python
def printWorld():
    yield("Hello World1")
    yield("Hello World2")
    yield("Hello World3")
```

```python
result=printWorld()
```

```python
print(next(result))
```

```python
# write a generator function that yields all prime nos between 1 to 50

def prime_no():
    for i in range(2,51):
        for j in range(2,i):
            if i%j==0:
                break
        else:
            yield(i,"prime")
        
ans=prime_no()
```

```python
print(next(ans))
```

```python
# write a generator function that yeilds cubes from 5 to 20

def findCube():
    for i in range(5,21):
        yield(i," cube :",i**3)
cubes=findCube()
```

```python
print(next(cubes))
```

```python
# 1 write a generator function that yields all armstrong numbers between 100 to 999


def armstrong_no():
    
    
    for i in range(100,1000):
        num=i
        total=0
        while num>0:
            last=num % 10 
            total+=last**3
            num=num//10
        if i==total:
            yield(i,"armstrong")
    
armstrong=armstrong_no()
```

```python
print(next(armstrong))
```
