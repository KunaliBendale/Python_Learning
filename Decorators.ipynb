```python
# decorator using arguments

def ATM(func):
    def insidefun(amt):
        print("Enter amount")
        func(amt)
        print("please collect receipt")
    return insidefun
```

```python
def intro(func):
    def wrapper(*args , **kwargs):
        print("hii")
        func(*args,**kwargs)
        print("Welcome to hefshine")
    return wrapper

@intro
def sample(*args,**kwargs):
    for data in args:
        print(data)
    for name,val in kwargs.items():
        print(name,val)
sample(56,78,24,78,34,name="kunali",education="MCA")
```

```python
# write a decoraTOR function validator() to validate age of user. if age >=18 "eligible to vote " otherwise "not eligible to vote

def validator(func):
    def wrapper(age):
        print("Your age is :" ,age)
        func()
        print("Hope you happy !!")
    return wrapper
@validator
def validate():
    age=int(input("Enter your age: "))
```

```python
def valid_statement(func):
    def InsideFunction(Name,roll):
        if isinstance(Name,str) and len(Name)<=8:
            func(Name,roll)
        else:
            print("invalid name")
    return InsideFunction

@valid_statement
def getDetails(Name,roll):
    print("Name is: ",Name)
    print("Roll no is: ",roll)
    
Name=input("ENter your name: ")
roll=int(input("ENter your roll no.: "))
getDetails(Name,roll)
```

```python
# Decorators as scheduler

import datetime
def scheduletask(func):
    def wrapper():
        hour1=datetime.datetime.now().hour
        if hour1>=10 and hour1<=18:
            func()
            print(hour1)
        else:
            print("Task is not allowed at this time")
    return wrapper

@scheduletask
def Test():
    print("Task is running")
```
