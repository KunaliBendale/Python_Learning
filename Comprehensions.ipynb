```python
# square each element from list1
list1=[1,2,24,34,576,7]
list2=[]

for i in list1:
    list2.append(i*i)
    
print(list2)
```

```python
#using list comprehension :

list1=[1,2,24,34,576,7]
[i*i for i in list1]
```

```python
# extract odd elements from list1
list1=[1,2,24,34,576,7]
list2=[]
for i in list1:
    if i%2!=0:
        list2.append(i)
print(list2)
```

```python
# using list comprehension

list1=[1,2,24,34,576,7]
[i for i in list1 if i %2 != 0]
```

```python
# list of squares of numbers from 1 to 10
[i*i for i in range(1,11)]
```

```python
# List of cubes of numbers from 1 to 20
[i**3 for i in range(1,21)]
```

```python
# create list of even numbers from 1 to 20
[i for i in range(1,21) if i % 2 ==0]
```

```python
# create list of odd numbers from 1 to 20
[i for i in range(1,21) if i % 2 !=0]
```

```python
#create list containing length of each word
list1=["python","java","html","css"]
[len(i) for i in list1]
```

```python
# convert all names in uppercase
list1=["python","java","html","css"]
[i.upper() for i in list1]
```

```python
#create list of numbers divisible by 5 from 1 to 100.
[i for i in range(1,101) if i % 5 ==0]
```

```python
# create list of first letters
fruits=["apple","Banana","Mango","Orange"]
[i[0] for i in fruits]
```

```python
#create list of positive nos.
list1=[2,4,67,-21,-45 -4,6,8,-9]
[i for i in list1 if i > 0]
```

```python

for i in range(2,50):
    for j in range(2,49):
        if i != j:
            if i % j == 0:
                print(f"{i} is not prime")
                break
    else:
        print(f"{i} is prime")
```
