```python
# using dict() constructor

College = dict(
    Name="Institute of Management and Research",
    Address="Jalgaon",
    Departments=["computer science","Pharma","Management"],
    NAAC_Accredited= True,
    Total_Admissions= 200,
    Reputation= True
)

print(College)
```

```python
# accessing element using key
print(College["Name"])

# updating element
College.update({"Address":"Pune"})
print(College)

# adding new element
College["pincode"]=425001
print(College)
```

```python
print(College.keys())

print(College.values())

print(College.items())
```

```python
#Write a Python function that takes a dictionary and a key, and returns the value if the key exists,
#else "Key not found".

def find_key(dictionary, key):
    if key in dictionary:
        return dictionary[key]
    else:
        return "Key not found"


print(find_key(Student, "name"))
print(find_key(Student, "marks"))
```

```python
#Write a program to remove the "city" key from the dictionary 
#person = {"name": "John", "age": 25, "city": "New York"}
#and print the resulting dictionary.

person = {"name": "John", "age": 25, "city": "New York"}
person.pop("city")
print(person)
```

```python
# swaps keys to value 
fruit_prices = {
    "apple": 50,
    "banana": 20,
    "orange": 30,
    "mango": 60,
    "grapes": 40
}

invert = {}

for key, value in fruit_prices.items():
    invert[value] = key

print(invert)
```
