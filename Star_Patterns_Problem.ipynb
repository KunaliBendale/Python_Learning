```python
# pyramid
for i in range(1,6):
    for j in range(1,i+1):
        print(i,end=" ")
    print()
```

```python
# reverse pyramid of nos.
for i in range(6,0,-1):
    for j in range(1,i+1):
        print(j,end="")
    print()
```
