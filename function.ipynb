```python
# check no is prime . if it is return Prime or return not prime

def is_prime(num):
    if num > 1:
        for i in range(2,num):
            if num % i == 0:
                return "not prime"
        else:
            return "Prime"
num = int(input("ENter number: "))
is_prime(num)
    
```

```python
# Write a function  find_largest(a, b, c)   that returns the largest of three numbers.

def find_largest(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else:
        return c
num1=int(input("ENter number 1: "))
num2=int(input("ENter number 2: "))
num3=int(input("ENter number 3: "))
print(find_largest(num1,num2,num3),"is the largest number. ")
```

```python
# Write a function   factorial(n)   that returns the factorial of a given number.
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact
factorial(5)
```

```python
# Write a function   count_vowels(text)   that returns the total number of vowels in a string.

def count_vowels(text):
    vowels="AEIOUaeiou"
    count=0
    for i in text:
        if i in vowels:
            count+=1
    return count     
print(count_vowels("Hello Kunali"))
```

```python
# Write a function   convert_celsius_to_fahrenheit(celsius)   that converts Celsius to Fahrenheit.

def convert_celcius_to_fahernheit(celcius):
    fahrenheit = (celcius * 9/5) + 32
    print(f"{celcius} celcius = {fahrenheit} fahrenheit")
    
convert_celcius_to_fahernheit(20)
```

```python
# Write a function   is_palindrome(text)   that checks whether a given string is a palindrome.

def is_palindrome(text):
    orig=text
    rev=""
    while len(orig)>0:
        last=orig[-1]
        rev= rev+last
        orig=orig[:-1]
        
    if text == rev:
        print("Palindrome")
    else:
        print("not palindrome")

is_palindrome("madam")

def chk_palindrome(text):
    if text[::-1] == text:
        print("Palindrome")
    else:
        print("not palindrome")
        
chk_palindrome("madam")
```
