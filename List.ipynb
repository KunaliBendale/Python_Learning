```python
#List

#collection of ordered elements
# mutable in nature
#can store duplicate elements
# create using square brackets , and list() constructor


# Methods to add elements in list

list1=[12,45,12,32,87,90,23,]
print(list1)

list1.append(48)
print(list1)

list1.insert(0,78)
print(list1)

list1.extend([23,47,68,99])
print(list1)
```

```python
# Methods to remove elements from list

list1.pop()         # removed last element by default
print(list1)

list1.pop(2)         # removed element of specific position
print(list1)

list1.remove(48)      # removed specific element
print(list1)
```

```python
print(list1.count(12))

list1.sort()    # by default sorts list in ascending order
print(list1)

list1.sort(reverse=True)
print(list1)

list2=list1.copy()
print(list2)

print(list1.index(23))

list1.reverse()
print(list1)

list1.clear()
print(list1)
```

```python
#Universal functions

list1=[23,54,67,89,45,90]
print("Maximum element: ",max(list1))
print("minumum element: ",min(list1))
print("length of list is: ",len(list1))
print("total : ",sum(list1))
```
