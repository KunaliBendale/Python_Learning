```python
#factorial using recursion

def factorial(n):
    if n ==0:                           # base case
        return 1
    else:
        return n * factorial(n-1)       # recursive call
    
print(factorial(6))
```

```python
# calclate sum of elements in list

def sum_of_list(lst):
    if len(lst)==0:
        return 0
    else:
        return lst[0] + sum_of_list(lst[1:])
    
print(sum_of_list([1,2,3,4,5]))
```

```python
# 2. Write a recursive function fibonacci(n) that returns the nth number in the Fibonacci sequence.
# Call the function with n = 6 and print the result.

def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    else:
        return fibonacci(n-1)+fibonacci(n-2)
    
print(fibonacci(6))    
```
